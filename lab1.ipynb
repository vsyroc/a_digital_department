# Загрузка модели

```python
import torch
print(torch.__version__)
print(torch.cuda.is_available())
```

```python
ner_model = build_model(configs.ner.ner_rus_bert, download=False)
```

# Пример текста

PER - персона, ORG - организация, LOC - локация. B - начало сущности, I- продолжение.

```python
text = "Илон Маск основал компанию Tesla в Калифорнии."

tokens, tags = ner_model([text])
tokens, tags = tokens[0], tags[0]

for t, tag in zip(tokens, tags):
    if tag != "O":
        print(f"{t} → {tag}")
```

# Несколько текстов

```python
texts = [
    "Владимир Путин встретился с Си Цзиньпином в Пекине.",
    "Google основали Ларри Пейдж и Сергей Брин в Стэнфорде.",
    "Достоевский написал Преступление и наказание в Петербурге.",
]

all_tokens, all_tags = ner_model(texts)

for i, text in enumerate(texts):
    print(f"\n{text}")
    for t, tag in zip(all_tokens[i], all_tags[i]):
        if tag != "O":
            print(f"  {t} → {tag}")
```
